```sql
-- Top 10 employees that have completed the most orders
SELECT TOP (10) SalespersonPersonID, COUNT(OrderID) AS CompletedOrders
FROM Sales.Orders
WHERE SalespersonPersonID IS NOT NULL 
GROUP BY SalespersonPersonID
ORDER BY CompletedOrders DESC;
```

```sql
-- All orders from a specific Sales person (SalespersonPersonID = 2)
SELECT SalespersonPersonID, OrderDate, OrderID, CustomerID
FROM Sales.Orders
WHERE SalespersonPersonID = 2
ORDER BY SalespersonPersonID, OrderDate DESC;
```

```sql
-- Orders completed by a specific Sales person for a specific customer (orders completed by SalespersonPersonID = 3 for CustomeriD = 28)
SELECT SalespersonPersonID, CustomerID, OrderDate, OrderID
FROM Sales.Orders
WHERE SalespersonPersonID = 3 AND CustomerID = 28
ORDER BY SalespersonPersonID, OrderDate DESC;
```

```sql
-- All orders made by a customer between 2 dates (orders made by CostumerID = 5 between 2014-06-30 and 2015-06-30)
SELECT CustomerID, OrderDate, OrderID, SalespersonPersonID
FROM Sales.Orders
WHERE CustomerID = 5 AND (OrderDate BETWEEN '2014-06-30' AND '2015-06-30')
ORDER BY CustomerID, OrderDate DESC;
```

```sql
-- Time between order and delivery date for each order for a specific sales person (SalespersonPersonID = 2)
SELECT SalespersonPersonID, CustomerID, OrderID, OrderDate, ExpectedDeliveryDate, 
    DATEDIFF(DAY, OrderDate, ExpectedDeliveryDate) AS ExpectedDaystoDeliver
FROM Sales.Orders
WHERE ExpectedDeliveryDate IS NOT NULL AND SalespersonPersonID = 2
ORDER BY CustomerID, OrderDate DESC;
```

```sql
-- Get the total number of orders placed on a specific date (2014-12-19)
SELECT COUNT(OrderID) AS TotalOrders
FROM Sales.Orders
WHERE CAST(OrderDate AS DATE) = '2014-12-19';
```

```sql
-- Top 10 customers that have placed the most orders
SELECT TOP (10) CustomerID, COUNT(OrderID) AS PlacedOrders
FROM Sales.Orders
WHERE CustomerID IS NOT NULL 
GROUP BY CustomerID
ORDER BY PlacedOrders DESC;
```

```sql
-- Top 10 dates that had the most orders placed
SELECT TOP (10) OrderDate, COUNT(OrderID) AS PlacedOrders
FROM Sales.Orders
WHERE OrderDate IS NOT NULL 
GROUP BY OrderDate
ORDER BY PlacedOrders DESC;
```

```sql
-- Customers who have placed more than 100 total orders
SELECT CustomerID, COUNT(OrderID) AS TotalOrders
FROM Sales.Orders
WHERE CustomerID IS NOT NULL 
GROUP BY CustomerID
HAVING COUNT(OrderID) > 100
ORDER BY TotalOrders DESC;
```

```sql
-- Customers who have placed less than 10 orders
SELECT CustomerID, COUNT(OrderID) AS TotalOrders
FROM Sales.Orders
WHERE CustomerID IS NOT NULL 
GROUP BY CustomerID
HAVING COUNT(OrderID) < 10
ORDER BY TotalOrders DESC;
```
