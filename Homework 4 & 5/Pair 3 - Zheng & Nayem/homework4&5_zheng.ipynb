```sql
WITH CustomerTotal AS (
    SELECT CustomerID, SUM(TotalDryItems + TotalChillerItems) AS TotalSpent
    FROM Sales.Invoices
    GROUP BY CustomerID
)
SELECT TOP 5 c.CustomerName, ct.TotalSpent
FROM CustomerTotal ct
JOIN Sales.Customers c ON c.CustomerID = ct.CustomerID
ORDER BY ct.TotalSpent DESC;
```

```sql
SELECT SupplierID, SupplierName
FROM Purchasing.Suppliers
WHERE SupplierID NOT IN (
    SELECT DISTINCT SupplierID
    FROM Purchasing.SupplierTransactions
);
```

```sql
SELECT CityName, COUNT(CustomerID) AS CustomerCount
FROM Application.Cities AS c
JOIN Sales.Customers AS cust ON c.CityID = cust.DeliveryCityID
GROUP BY CityName
HAVING COUNT(CustomerID) > 0;
```

```sql
SELECT InvoiceID, COUNT(*) AS NumberOfItems
FROM Sales.InvoiceLines
GROUP BY InvoiceID
HAVING COUNT(*) > (
    SELECT AVG(ItemCount)
    FROM (
        SELECT COUNT(*) AS ItemCount
        FROM Sales.InvoiceLines
        GROUP BY InvoiceID
    ) AS AvgItems
);
```

```sql
SELECT s.SupplierID, s.SupplierName, COUNT(po.PurchaseOrderID) AS NumPurchaseOrders
FROM Purchasing.Suppliers s
JOIN Purchasing.PurchaseOrders po ON s.SupplierID = po.SupplierID
GROUP BY s.SupplierID, s.SupplierName
HAVING COUNT(po.PurchaseOrderID) > 5
ORDER BY NumPurchaseOrders DESC;
```

```sql
WITH ProductCounts AS (
    SELECT po.SupplierID, COUNT(DISTINCT pol.StockItemID) AS DistinctProducts
    FROM Purchasing.PurchaseOrders po
    JOIN Purchasing.PurchaseOrderLines pol ON po.PurchaseOrderID = pol.PurchaseOrderID
    GROUP BY po.SupplierID
)
SELECT TOP 1 s.SupplierName, pc.DistinctProducts
FROM ProductCounts pc
JOIN Purchasing.Suppliers s ON pc.SupplierID = s.SupplierID
ORDER BY pc.DistinctProducts DESC;
```

```sql
SELECT TOP 5 c.CustomerID, c.CustomerName, COUNT(ct.CustomerTransactionID) AS NumTransactions
FROM Sales.Customers c
JOIN Sales.CustomerTransactions ct ON c.CustomerID = ct.CustomerID
GROUP BY c.CustomerID, c.CustomerName
ORDER BY NumTransactions DESC;
```

```sql
WITH DailyInvoices AS (
    SELECT InvoiceDate, COUNT(InvoiceID) AS InvoiceCount
    FROM Sales.Invoices
    GROUP BY InvoiceDate
),
MaxDay AS (
    SELECT TOP 1 InvoiceDate
    FROM DailyInvoices
    ORDER BY InvoiceCount DESC
)
SELECT *
FROM Sales.Invoices
WHERE InvoiceDate = (SELECT InvoiceDate FROM MaxDay);
```

```sql
WITH CustomerSpent AS (
    SELECT CustomerID, SUM(TotalDryItems + TotalChillerItems) AS TotalSpent
    FROM Sales.Invoices
    GROUP BY CustomerID
),
AvgSpent AS (
    SELECT AVG(TotalSpent) AS AverageSpent
    FROM CustomerSpent
)
SELECT c.CustomerName, cs.TotalSpent
FROM CustomerSpent cs
JOIN Sales.Customers c ON cs.CustomerID = c.CustomerID
WHERE cs.TotalSpent > (SELECT AverageSpent FROM AvgSpent);
```

```sql
WITH MethodUsage AS (
    SELECT DeliveryMethodID, COUNT(*) AS UsageCount
    FROM Sales.Invoices
    GROUP BY DeliveryMethodID
)
SELECT TOP 1 dm.DeliveryMethodName, mu.UsageCount
FROM MethodUsage mu
JOIN Application.DeliveryMethods dm ON mu.DeliveryMethodID = dm.DeliveryMethodID
ORDER BY mu.UsageCount DESC;
```
