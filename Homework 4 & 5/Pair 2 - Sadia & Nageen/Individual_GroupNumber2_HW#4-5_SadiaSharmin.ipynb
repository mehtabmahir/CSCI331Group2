```sql
--1) Find the cheapest stock item
--Uses subquery to first select the min unit price
SELECT StockItemID, StockItemName, UnitPrice
FROM Warehouse.StockItems
WHERE UnitPrice = 
    (SELECT MIN(UnitPrice) 
    FROM Warehouse.StockItems);
```

```sql
--2) Gets the employees that have processed more than 5000 orders
--Uses a CTE
WITH TotalOrdersPerEmployee AS 
    (SELECT SalespersonPersonID, COUNT(*) AS OrdersProcessed
    FROM Sales.Orders
    GROUP BY SalespersonPersonID)
SELECT * FROM TotalOrdersPerEmployee WHERE OrdersProcessed > 5000
ORDER BY OrdersProcessed DESC;
```

```sql
--3) Gets customers that have placed more than 100 orders
--Uses CTE
WITH CustomerTotalOrders AS 
    (SELECT CustomerID, COUNT(OrderID) AS TotalOrders
    FROM Sales.Orders
    GROUP BY CustomerID)
SELECT CustomerID, TotalOrders
FROM CustomerTotalOrders
WHERE TotalOrders > 100
ORDER BY TotalOrders DESC;
```

```sql
-- 4) Customers with outstanding balances greater than 1000
--Uses CTE 
--Inner join to combine tables
WITH CustomerBalance AS 
    (SELECT CustomerID, SUM(OutstandingBalance) AS TotalBalance
    FROM Sales.CustomerTransactions
    GROUP BY CustomerID)
SELECT c.CustomerID, b.TotalBalance
FROM CustomerBalance b
INNER JOIN Sales.Customers c ON b.CustomerID = c.CustomerID
WHERE b.TotalBalance > 1000
ORDER BY TotalBalance DESC;
```

```sql
--5) Find the first order placed by each customer
-- uses subquery
SELECT o.CustomerID, o.OrderDate
FROM Sales.Orders AS o
WHERE OrderDate = 
    (SELECT MIN(OrderDate) 
    FROM Sales.Orders 
    WHERE CustomerID = o.CustomerID);
```

```sql
--6) Get total number of orders placed by each customer
--uses CTE
WITH OrderCount AS 
    (SELECT CustomerID, COUNT(OrderID) AS Total
    FROM Sales.Orders
    GROUP BY CustomerID)
SELECT c.CustomerID, c.CustomerName, o.Total
FROM Sales.Customers AS c
INNER JOIN OrderCount o ON c.CustomerID = o.CustomerID
ORDER BY Total DESC;
```

```sql
--7) Adds to last query, customers with less than 10 total orders 
--Uses CTE
WITH OrderCount AS 
    (SELECT CustomerID, COUNT(OrderID) AS Total
    FROM Sales.Orders
    GROUP BY CustomerID)
SELECT c.CustomerID, c.CustomerName, o.Total
FROM Sales.Customers AS c
INNER JOIN OrderCount o ON c.CustomerID = o.CustomerID
WHERE Total < 10
ORDER BY TOTAL DESC;
```

```sql
--8) Get all orders placed on 03/14/2013
--uses subquery
SELECT o.CustomerID, o.OrderDate
FROM Sales.Orders AS o
WHERE o.OrderDate IN 
    (SELECT OrderDate
     FROM Sales.Orders
     WHERE OrderDate = '2013-03-14');
```

```sql
--9) Finds most expensive stock item
--uses CTE and subquery
WITH MostExpensiveItem AS 
    (SELECT MAX(UnitPrice) AS MaxPrice
    FROM Warehouse.StockItems)
SELECT s.StockItemID, s.StockItemName, s.UnitPrice
FROM Warehouse.StockItems AS s
WHERE UnitPrice = 
    (SELECT MaxPrice 
    FROM MostExpensiveItem);
```

```sql
--10) Find stock items that has a weight of 5
--uses subquery
SELECT w.StockItemID, w.StockItemName, w.TypicalWeightPerUnit
FROM Warehouse.StockItems AS w
WHERE w.TypicalWeightPerUnit IN 
    (SELECT TypicalWeightPerUnit
     FROM Warehouse.StockItems
     WHERE TypicalWeightPerUnit = 5);
```
