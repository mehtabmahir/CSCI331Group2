## Step 3: Create the Star Schema Tables

I created the base versions of the dimension and fact tables, each with a non-IDENTITY primary key and the required audit columns.

```sql
CREATE TABLE [CH01-01-Dimension].[DimCustomer] (
    CustomerKey INT PRIMARY KEY,
    CustomerName NVARCHAR(100),
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);

CREATE TABLE [CH01-01-Dimension].[DimProduct] (
    ProductKey INT PRIMARY KEY,
    ProductName NVARCHAR(100),
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);

CREATE TABLE [CH01-01-Dimension].[DimTerritory] (
    TerritoryKey INT PRIMARY KEY,
    TerritoryName NVARCHAR(100),
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);

CREATE TABLE [CH01-01-Dimension].[DimOccupation] (
    OccupationKey INT PRIMARY KEY,
    OccupationName NVARCHAR(100),
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);

CREATE TABLE [CH01-01-Fact].[Data] (
    DataKey INT PRIMARY KEY,
    SomeMeasure INT,
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);
```

## Step 6: Create Sequence Objects for All Tables

I created sequences for all the dimension and fact tables to replace identity usage.

```sql
CREATE SCHEMA PkSequence;
GO

CREATE SEQUENCE PkSequence.DimCustomerSequenceObject
    START WITH 1
    INCREMENT BY 1;
GO

CREATE SEQUENCE PkSequence.DimProductSequenceObject
    START WITH 1
    INCREMENT BY 1;
GO

CREATE SEQUENCE PkSequence.DimTerritorySequenceObject
    START WITH 1
    INCREMENT BY 1;
GO

CREATE SEQUENCE PkSequence.DimOccupationSequenceObject
    START WITH 1
    INCREMENT BY 1;
GO

CREATE SEQUENCE PkSequence.DimProductCategorySequenceObject
    START WITH 1
    INCREMENT BY 1;
GO

CREATE SEQUENCE PkSequence.DimProductSubcategorySequenceObject
    START WITH 1
    INCREMENT BY 1;
GO

CREATE SEQUENCE PkSequence.FactDataSequenceObject
    START WITH 1
    INCREMENT BY 1;
GO
```

##  Step 7: Stored Procedures to Load Dimension Tables

I created procedures to load DimCustomer, DimProduct, and DimTerritory.

```sql
-- LoadDimCustomer
CREATE PROCEDURE [CH01-01-Dimension].[LoadDimCustomer]
AS
BEGIN
    INSERT INTO [CH01-01-Dimension].[DimCustomer] (
        CustomerKey,
        CustomerName,
        UserAuthorizationKey,
        DateAdded,
        DateOfLastUpdate
    )
    SELECT
        NEXT VALUE FOR PkSequence.DimCustomerSequenceObject,
        Source.CustomerName,
        1,
        SYSDATETIME(),
        SYSDATETIME()
    FROM [FileUpload].[OriginallyLoadedData] AS Source;
END;
GO

-- LoadDimProduct
CREATE PROCEDURE [CH01-01-Dimension].[LoadDimProduct]
AS
BEGIN
    INSERT INTO [CH01-01-Dimension].[DimProduct] (
        ProductKey,
        ProductName,
        UserAuthorizationKey,
        DateAdded,
        DateOfLastUpdate
    )
    SELECT
        NEXT VALUE FOR PkSequence.DimProductSequenceObject,
        Source.ProductName,
        1,
        SYSDATETIME(),
        SYSDATETIME()
    FROM [FileUpload].[OriginallyLoadedData] AS Source;
END;
GO

-- LoadDimTerritory
CREATE PROCEDURE [CH01-01-Dimension].[LoadDimTerritory]
AS
BEGIN
    INSERT INTO [CH01-01-Dimension].[DimTerritory] (
        TerritoryKey,
        TerritoryName,
        UserAuthorizationKey,
        DateAdded,
        DateOfLastUpdate
    )
    SELECT
        NEXT VALUE FOR PkSequence.DimTerritorySequenceObject,
        Source.TerritoryName,
        1,
        SYSDATETIME(),
        SYSDATETIME()
    FROM [FileUpload].[OriginallyLoadedData] AS Source;
END;
GO
```
