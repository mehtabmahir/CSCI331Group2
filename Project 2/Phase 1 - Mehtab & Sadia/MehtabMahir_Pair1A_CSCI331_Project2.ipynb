## Step 2: Create the Required Schema

The schemas `CH01-01-Dimension`, `CH01-01-Fact`, and `DbSecurity` did not exist, so I created them:

```sql
CREATE SCHEMA [CH01-01-Dimension];
GO

CREATE SCHEMA [CH01-01-Fact];
GO

CREATE SCHEMA [DbSecurity];
GO
```

## Step 3: Create the Star Schema Tables

I created the base versions of the dimension and fact tables, each with a non-IDENTITY primary key and the required audit columns.

```sql
CREATE TABLE [CH01-01-Dimension].[DimCustomer] (
    CustomerKey INT PRIMARY KEY,
    CustomerName NVARCHAR(100),
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);

CREATE TABLE [CH01-01-Dimension].[DimProduct] (
    ProductKey INT PRIMARY KEY,
    ProductName NVARCHAR(100),
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);

CREATE TABLE [CH01-01-Dimension].[DimTerritory] (
    TerritoryKey INT PRIMARY KEY,
    TerritoryName NVARCHAR(100),
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);

CREATE TABLE [CH01-01-Dimension].[DimOccupation] (
    OccupationKey INT PRIMARY KEY,
    OccupationName NVARCHAR(100),
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);

CREATE TABLE [CH01-01-Fact].[Data] (
    DataKey INT PRIMARY KEY,
    SomeMeasure INT,
    UserAuthorizationKey INT NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME(),
    DateOfLastUpdate DATETIME2 DEFAULT SYSDATETIME()
);
```

##  Step 4: Create the `DbSecurity.UserAuthorization` Table

```sql
CREATE TABLE [DbSecurity].[UserAuthorization] (
    UserAuthorizationKey INT PRIMARY KEY,
    ClassTime NCHAR(5),
    [Individual project] NVARCHAR(60) DEFAULT 'PROJECT 2 RECREATE THE BICLASS DATABASE STAR SCHEMA',
    GroupMemberLastName NVARCHAR(35) NOT NULL,
    GroupMemberFirstName NVARCHAR(25) NOT NULL,
    GroupName NVARCHAR(20) NOT NULL,
    DateAdded DATETIME2 DEFAULT SYSDATETIME()
);
```

##  Step 5: Back Up the Database (WSL-Compatible)

Since I am using SQL Server in WSL, I backed up the database to a location and then copied it to Windows using docker cp.
