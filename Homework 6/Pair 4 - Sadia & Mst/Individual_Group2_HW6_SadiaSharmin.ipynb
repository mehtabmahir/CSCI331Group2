```sql
--1. Get all product names and the product descriptions
--use union all so it gets all rows, not just distinct ones
SELECT Name FROM Production.Product
UNION ALL 
SELECT Description FROM Production.ProductDescription
```

```sql
--2. Get product ids that are in both the producton.product and sales.specialofferproduct tables
SELECT ProductID FROM Production.Product  
INTERSECT  
SELECT ProductID FROM Sales.SpecialOfferProduct;
```

```sql
--3 find all businessentityids that do not exist in jobcandidateid
SELECT BusinessEntityID FROM HumanResources.Employee
EXCEPT 
SELECT JobCandidateID FROM HumanResources.JobCandidate
```

```sql
--4 get all the employee ids and the shift ids
--uses union all, so takes all rows, not just distinct
SELECT BusinessEntityID FROM HumanResources.Employee
UNION ALL 
SELECT ShiftID FROM HumanResources.EmployeeDepartmentHistory
```

```sql
--5 get all employee ids from humanresources.employee that are not in the sales.salesperson table
SELECT BusinessEntityID FROM HumanResources.Employee
EXCEPT
SELECT BusinessEntityID FROM Sales.SalesPerson;
```

```sql
--6 Get distinct product ids from production.product and production.productreview
SELECT ProductID FROM Production.Product
UNION 
SELECT ProductID FROM Production.ProductReview;
```

```sql
--7 get product ids in production but not in the sales table
SELECT ProductID FROM Production.ProductInventory
EXCEPT
SELECT ProductID FROM Sales.SalesOrderDetail;
```

```sql
--8 get all order and ship dates together
SELECT OrderDate FROM Sales.SalesOrderHeader
UNION
SELECT ShipDate FROM Sales.SalesOrderHeader;
```

```sql
--9 get hashed password and email address of each person
SELECT PasswordHash FROM Person.Password
UNION ALL
SELECT EmailAddress FROM Person.EmailAddress;
```

```sql
--10 get all address ids and state province ids
SELECT AddressID FROM Person.BusinessEntityAddress
UNION ALL
SELECT StateProvinceID FROM Person.StateProvince;
```
