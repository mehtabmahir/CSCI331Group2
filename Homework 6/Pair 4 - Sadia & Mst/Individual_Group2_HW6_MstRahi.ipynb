```sql
USE AdventureWorks2019;

-- Proposition 1: Products that exist in inventory but have never been ordered.
-- Helps identify obsolete or slow-moving stock.
SELECT ProductID 
FROM Production.ProductInventory
EXCEPT
SELECT ProductID 
FROM Sales.SalesOrderDetail;
```

```sql
USE AdventureWorks2019;

-- Proposition 2: Employees who have worked in more than one department and are also in the Sales Team.
-- Helps track internal mobility within the company.

SELECT BusinessEntityID 
FROM HumanResources.EmployeeDepartmentHistory
INTERSECT
SELECT BusinessEntityID 
FROM Sales.SalesPerson;
```

```sql
USE AdventureWorks2019;

-- Proposition 3: Customers who placed orders in 2011 but not in 2012.
-- Useful for tracking customer retention and repeat purchases.

SELECT CustomerID 
FROM Sales.SalesOrderHeader
WHERE YEAR(OrderDate) = 2011
EXCEPT
SELECT CustomerID 
FROM Sales.SalesOrderHeader
WHERE YEAR(OrderDate) = 2012;
```

```sql
USE AdventureWorks2019;

-- Proposition 4: Products that have been ordered and also appear in special offers.
-- Helps marketing analyze how well promotions drive actual purchases.

SELECT ProductID
FROM Sales.SalesOrderDetail
INTERSECT
SELECT ProductID 
FROM Sales.SpecialOfferProduct;
```

```sql
USE AdventureWorks2019;

-- Proposition 5: Find salespeople who made sales in both 2012 and 2013.
-- Uses INTERSECT to identify common SalesPersonID between the two years.

SELECT SalesPersonID
FROM Sales.SalesOrderHeader
WHERE OrderDate >= '2013-01-01' AND OrderDate < '2014-01-01'
INTERSECT
SELECT SalesPersonID
FROM Sales.SalesOrderHeader
WHERE OrderDate >= '2014-01-01' AND OrderDate < '2015-01-01';
```

```sql
USE AdventureWorks2019;

-- Proposition 6: Customers who placed orders both online and in physical stores.
-- Identifies multi-channel shoppers for targeted marketing.
SELECT CustomerID 
FROM Sales.SalesOrderHeader
WHERE OnlineOrderFlag = 1
INTERSECT
SELECT CustomerID 
FROM Sales.SalesOrderHeader
WHERE OnlineOrderFlag = 0;
```

```sql
USE AdventureWorks2019;

-- Proposition 7: Vendors who have supplied products but do not have active contracts.
-- Useful for supplier management and contract reviews.
SELECT VendorID 
FROM Purchasing.PurchaseOrderHeader
EXCEPT
SELECT BusinessEntityID 
FROM Purchasing.Vendor
WHERE ActiveFlag = 1;
```

```sql
USE AdventureWorks2019;

-- Proposition 8: Products that were discounted but never purchased.
-- Helps evaluate ineffective promotions.
SELECT ProductID 
FROM Sales.SpecialOfferProduct
EXCEPT
SELECT ProductID 
FROM Sales.SalesOrderDetail;
```

```sql
USE AdventureWorks2019;

-- Proposition 9: Vendors who previously supplied products but did not supply any in the last year.
-- Helps identify inactive vendors.

SELECT DISTINCT VendorID
FROM Purchasing.PurchaseOrderHeader
WHERE YEAR(OrderDate) < YEAR(GETDATE()) - 1
EXCEPT
SELECT DISTINCT VendorID
FROM Purchasing.PurchaseOrderHeader
WHERE YEAR(OrderDate) = YEAR(GETDATE()) - 1;
```

```sql
USE AdventureWorks2019;

-- Proposition 10: Employees who were assigned a sales territory and also made at least one sale.
-- Helps validate that assigned salespeople are actively making sales.

SELECT BusinessEntityID
FROM Sales.SalesPerson
WHERE TerritoryID IS NOT NULL
INTERSECT
SELECT DISTINCT SalesPersonID
FROM Sales.SalesOrderHeader
WHERE SalesPersonID IS NOT NULL;
```
