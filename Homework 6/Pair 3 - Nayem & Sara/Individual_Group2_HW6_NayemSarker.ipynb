### **Proposition 1: List All Distinct Product Names Sold or Discounted**

```sql
USE AdventureWorks2019;
GO

-- Products sold
SELECT DISTINCT p.Name
FROM Sales.SalesOrderDetail sod
JOIN Production.Product p ON sod.ProductID = p.ProductID

UNION

-- Products in special offers
SELECT DISTINCT p.Name
FROM Sales.SpecialOfferProduct sop
JOIN Production.Product p ON sop.ProductID = p.ProductID;
```

## **Proposition 2: Get All Product Names That Were Both Sold and Discounted**

```sql
USE AdventureWorks2019;
GO

SELECT DISTINCT p.Name
FROM Sales.SalesOrderDetail sod
JOIN Production.Product p ON sod.ProductID = p.ProductID

INTERSECT

SELECT DISTINCT p.Name
FROM Sales.SpecialOfferProduct sop
JOIN Production.Product p ON sop.ProductID = p.ProductID;
```

## **Proposition 3: Find All Product Names That Were Discounted but Never Sold**

```sql
USE AdventureWorks2019;
GO

SELECT DISTINCT p.Name
FROM Sales.SpecialOfferProduct sop
JOIN Production.Product p ON sop.ProductID = p.ProductID

EXCEPT

SELECT DISTINCT p.Name
FROM Sales.SalesOrderDetail sod
JOIN Production.Product p ON sod.ProductID = p.ProductID;
```

## **Proposition 4: List All Territories That Have Either a Salesperson or Customer**

```sql
USE AdventureWorks2019;
GO

SELECT TerritoryID
FROM Sales.SalesPerson
WHERE TerritoryID IS NOT NULL

UNION

SELECT TerritoryID
FROM Sales.Customer
WHERE TerritoryID IS NOT NULL;
```

## **Proposition 5: Find Territories Shared by Both Salespeople and Customers**

```sql
USE AdventureWorks2019;
GO

SELECT TerritoryID
FROM Sales.SalesPerson
WHERE TerritoryID IS NOT NULL

INTERSECT

SELECT TerritoryID
FROM Sales.Customer
WHERE TerritoryID IS NOT NULL;
```

## **Proposition 6: Find Vendors Who Supply Products, but Their Products Have Never Been Sold**

```sql
USE AdventureWorks2019;
GO

-- Vendors with products listed in the catalog
SELECT DISTINCT v.Name
FROM Purchasing.ProductVendor pv
JOIN Purchasing.Vendor v ON pv.BusinessEntityID = v.BusinessEntityID

EXCEPT

-- Vendors whose products were sold
SELECT DISTINCT v.Name
FROM Purchasing.ProductVendor pv
JOIN Purchasing.Vendor v ON pv.BusinessEntityID = v.BusinessEntityID
JOIN Sales.SalesOrderDetail sod ON pv.ProductID = sod.ProductID;
```

## **Proposition 7: List All Employee Job Titles and Customer Occupations**

```sql
USE AdventureWorks2019;
GO

SELECT JobTitle AS Title
FROM HumanResources.Employee

UNION ALL

SELECT Title
FROM Person.Person;

```

## **Proposition 8: Get Cities Where Employees and Customers Both Reside**

```sql
USE AdventureWorks2019;
GO

SELECT City
FROM Person.BusinessEntityAddress bea
JOIN Person.Address a ON bea.AddressID = a.AddressID
JOIN HumanResources.Employee e ON bea.BusinessEntityID = e.BusinessEntityID

INTERSECT

SELECT City
FROM Person.BusinessEntityAddress bea
JOIN Person.Address a ON bea.AddressID = a.AddressID
JOIN Sales.Customer c ON bea.BusinessEntityID = c.CustomerID;
```

## **Proposition 9: List Cities Where Customers Reside but No Employees Do**

```sql
USE AdventureWorks2019;
GO

SELECT City
FROM Person.BusinessEntityAddress bea
JOIN Person.Address a ON bea.AddressID = a.AddressID
JOIN Sales.Customer c ON bea.BusinessEntityID = c.CustomerID

EXCEPT

SELECT City
FROM Person.BusinessEntityAddress bea
JOIN Person.Address a ON bea.AddressID = a.AddressID
JOIN HumanResources.Employee e ON bea.BusinessEntityID = e.BusinessEntityID;
```

## **Proposition 10: Return Cities in Employee List First, Then Customer List (Ordered Output)**

```sql
USE AdventureWorks2019;
GO

SELECT City, 'Employee' AS Source
FROM Person.BusinessEntityAddress bea
JOIN Person.Address a ON bea.AddressID = a.AddressID
JOIN HumanResources.Employee e ON bea.BusinessEntityID = e.BusinessEntityID

UNION ALL

SELECT City, 'Customer' AS Source
FROM Person.BusinessEntityAddress bea
JOIN Person.Address a ON bea.AddressID = a.AddressID
JOIN Sales.Customer c ON bea.BusinessEntityID = c.CustomerID

ORDER BY Source, City;
```
