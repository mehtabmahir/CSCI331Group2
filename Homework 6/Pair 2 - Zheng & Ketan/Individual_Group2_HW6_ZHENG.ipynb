```sql
SELECT g.EnglishCountryRegionName AS country
FROM dbo.DimEmployee e
JOIN dbo.DimGeography g ON e.SalesTerritoryKey = g.SalesTerritoryKey
UNION
SELECT g.EnglishCountryRegionName AS country
FROM dbo.DimCustomer c
JOIN dbo.DimGeography g ON c.GeographyKey = g.GeographyKey;

```

```sql
SELECT g.City
FROM dbo.DimEmployee e
JOIN dbo.DimGeography g ON e.SalesTerritoryKey = g.SalesTerritoryKey
INTERSECT
SELECT g.City
FROM dbo.DimCustomer c
JOIN dbo.DimGeography g ON c.GeographyKey = g.GeographyKey;

```

```sql
SELECT g.StateProvinceName
FROM dbo.DimEmployee e
JOIN dbo.DimGeography g ON e.SalesTerritoryKey = g.SalesTerritoryKey
EXCEPT
SELECT g.StateProvinceName
FROM dbo.DimCustomer c
JOIN dbo.DimGeography g ON c.GeographyKey = g.GeographyKey;
```

```sql
SELECT g.EnglishCountryRegionName AS country, g.StateProvinceName AS state_province, g.City
FROM dbo.DimEmployee e
JOIN dbo.DimGeography g ON e.SalesTerritoryKey = g.SalesTerritoryKey
UNION ALL
SELECT g.EnglishCountryRegionName AS country, g.StateProvinceName AS state_province, g.City
FROM dbo.DimCustomer c
JOIN dbo.DimGeography g ON c.GeographyKey = g.GeographyKey;

```

```sql
SELECT g.EnglishCountryRegionName AS country
FROM dbo.DimEmployee e
JOIN dbo.DimGeography g ON e.SalesTerritoryKey = g.SalesTerritoryKey
INTERSECT
SELECT g.EnglishCountryRegionName AS country
FROM dbo.DimCustomer c
JOIN dbo.DimGeography g ON c.GeographyKey = g.GeographyKey
INTERSECT
SELECT g.EnglishCountryRegionName AS country
FROM dbo.DimReseller r
JOIN dbo.DimGeography g ON r.GeographyKey = g.GeographyKey;
```

```sql
(
    SELECT g.EnglishCountryRegionName AS country, g.StateProvinceName AS state_province, g.City
    FROM dbo.DimEmployee e
    JOIN dbo.DimGeography g ON e.SalesTerritoryKey = g.SalesTerritoryKey
    UNION
    SELECT g.EnglishCountryRegionName AS country, g.StateProvinceName AS state_province, g.City
    FROM dbo.DimReseller r
    JOIN dbo.DimGeography g ON r.GeographyKey = g.GeographyKey
)
EXCEPT
SELECT g.EnglishCountryRegionName AS country, g.StateProvinceName AS state_province, g.City
FROM dbo.DimCustomer c
JOIN dbo.DimGeography g ON c.GeographyKey = g.GeographyKey;
```

```sql
SELECT SalesTerritoryKey AS territory_id, SalesOrderNumber, OrderDate
FROM (SELECT TOP 2 SalesTerritoryKey, SalesOrderNumber, OrderDate
      FROM dbo.FactInternetSales
      WHERE SalesTerritoryKey = 1
      ORDER BY OrderDate DESC, SalesOrderNumber DESC) AS territory1_orders
UNION ALL
SELECT SalesTerritoryKey AS territory_id, SalesOrderNumber, OrderDate
FROM (SELECT TOP 2 SalesTerritoryKey, SalesOrderNumber, OrderDate
      FROM dbo.FactInternetSales
      WHERE SalesTerritoryKey = 2
      ORDER BY OrderDate DESC, SalesOrderNumber DESC) AS territory2_orders;
```

```sql
SELECT CustomerKey AS customer_id, SalesOrderNumber, OrderDate
FROM (SELECT TOP 2 CustomerKey, SalesOrderNumber, OrderDate
      FROM dbo.FactInternetSales
      WHERE CustomerKey = 11000
      ORDER BY OrderDate DESC, SalesOrderNumber DESC) AS customer1_orders
UNION ALL
SELECT CustomerKey AS customer_id, SalesOrderNumber, OrderDate
FROM (SELECT TOP 2 CustomerKey, SalesOrderNumber, OrderDate
      FROM dbo.FactInternetSales
      WHERE CustomerKey = 11001
      ORDER BY OrderDate DESC, SalesOrderNumber DESC) AS customer2_orders;
```
