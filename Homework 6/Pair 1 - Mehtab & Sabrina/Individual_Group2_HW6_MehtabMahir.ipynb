Query 1: Combine Product Names Using UNION
Proposition: Retrieve a unique list of product names from the Production.Product and Production.ProductModel tables.

```sql
SELECT Name
FROM Production.Product
UNION
SELECT Name
FROM Production.ProductModel;
```

Query 2: Combine Product Names Using UNION ALL
Proposition: Retrieve all product names (including duplicates) from the two tables.

```sql
SELECT Name
FROM Production.Product
UNION ALL
SELECT Name
FROM Production.ProductModel;
```

Query 3: Find Common IDs Between Person.Person and Sales.Customer Using INTERSECT
Proposition: Identify the IDs that appear in both Person.Person (BusinessEntityID) and Sales.Customer (PersonID).

```sql
SELECT BusinessEntityID
FROM Person.Person
INTERSECT
SELECT PersonID
FROM Sales.Customer
WHERE PersonID IS NOT NULL;
```

Query 4: Find Persons Who Are Not Customers Using EXCEPT
Proposition: List BusinessEntityIDs from Person.Person that do not appear as PersonID in Sales.Customer.

```sql
SELECT BusinessEntityID
FROM Person.Person
EXCEPT
SELECT PersonID
FROM Sales.Customer
WHERE PersonID IS NOT NULL;
```

Query 5: Find Common Cities in Addresses and Shipping Addresses Using INTERSECT
Proposition: Determine which cities appear both in the Person.Address table and as shipping addresses in Sales.SalesOrderHeader.

```sql
SELECT City
FROM Person.Address
INTERSECT
SELECT A.City
FROM Sales.SalesOrderHeader AS SOH
JOIN Person.Address AS A ON SOH.ShipToAddressID = A.AddressID;
```

Query 6: Combine Customers and Vendors Using UNION
Proposition: Create a distinct list of entity IDs with a label indicating 'Customer' or 'Vendor' by combining Sales.Customer and Purchasing.Vendor.
Note: For vendors, the correct column is BusinessEntityID.

```sql
SELECT CustomerID AS EntityID, 'Customer' AS EntityType
FROM Sales.Customer
UNION
SELECT BusinessEntityID AS EntityID, 'Vendor' AS EntityType
FROM Purchasing.Vendor;
```

Query 7: Combine Customers and Vendors Using UNION ALL
Proposition: List all entity IDs with their type from both tables without eliminating duplicates.
Note: For vendors, using BusinessEntityID.

```sql
SELECT CustomerID AS EntityID, 'Customer' AS EntityType
FROM Sales.Customer
UNION ALL
SELECT BusinessEntityID AS EntityID, 'Vendor' AS EntityType
FROM Purchasing.Vendor;
```

Query 8: Find Vendors Not Listed as Customers Using EXCEPT
Proposition: Identify Vendor IDs (using BusinessEntityID) from Purchasing.Vendor that do not appear in Sales.Customer as CustomerID.

```sql
SELECT BusinessEntityID
FROM Purchasing.Vendor
EXCEPT
SELECT CustomerID
FROM Sales.Customer;
```

Query 9: Retrieve Products with a High List Price and a Defined Color Using INTERSECT
Proposition: Return products from Production.Product that have a ListPrice greater than 100 and a non-null Color.

```sql
SELECT ProductID, Name, ListPrice
FROM Production.Product
WHERE ListPrice > 100
INTERSECT
SELECT ProductID, Name, ListPrice
FROM Production.Product
WHERE Color IS NOT NULL;
```

Query 10: Find Products Priced Above 100 But Not Over 500 Using EXCEPT
Proposition: List products with ListPrice above 100 while excluding those with ListPrice over 500.

```sql
SELECT ProductID, Name, ListPrice
FROM Production.Product
WHERE ListPrice > 100
EXCEPT
SELECT ProductID, Name, ListPrice
FROM Production.Product
WHERE ListPrice > 500;
```
