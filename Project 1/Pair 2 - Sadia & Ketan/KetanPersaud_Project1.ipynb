# Case 1:

On the faithful day of 2009-04-15 a fire broke out in a building full of workers at ADVENTUREWORKS2019. They were all safely evacuated. At the scene officers and detectives suggest that there was foul play. Based off eye witnesses, the description given was a man in his 40s and had no ring on his finger.

```sql
USE AdventureWorks2019;

SELECT P.FirstName, P.LastName, E.JobTitle, EA.EmailAddress, PA.PasswordHash 
FROM HumanResources.Employee as E
INNER JOIN Person.EmailAddress as EA 
    ON E.BusinessEntityID = EA.BusinessEntityID
INNER JOIN Person.[Password] as PA
    ON E.BusinessEntityID = PA.BusinessEntityID
INNER JOIN Person.Person as P
    ON E.BusinessEntityID = P.BusinessEntityID
WHERE BirthDate < '1969-04-15' AND BirthDate > '1959-04-15' AND HireDate > '2009-04-15' AND Gender = 'M' AND MaritalStatus = 'S'
```

Password hash 

<span style="color: var(--vscode-foreground);">Why he was there</span><span style="color: var(--vscode-foreground);">. (Job)</span>

# Case 2

It has been discovered that in WideWorldImporters there has been an sales person that has been pocketing any order made in January of 2014. It has been made public that it is his ID is 14. We must find these people he has taken from and what he has taken from them.

```sql
USE WideWorldImporters;

SELECT O.CustomerID, OL.StockItemID
FROM Sales.Orders as O
INNER JOIN Sales.OrderLines as OL
    ON O.OrderID = OL.OrderID
WHERE MONTH(O.OrderDate) = 01 AND YEAR(O.OrderDate) = 2014 AND O.SalespersonPersonID = 14
```

# Objective:

find the customer ID that Sales person 14 stole from

find the item ID that was stolen

# Schema Needed:

Sales.Orders

Sales.OrderLines

# NOTES:

the sales person ID is 14

the month and year is given

# Case 4

Rumor has it that the Vice President of Engineering has starting dating another employee from AdventureWorks2019. She has give this person a rather large bonus because of it. We must find out who this is.

```sql
USE AdventureWorks2019;
SELECT TOP(1) P.FirstName, P.LastName, E.JobTitle, SP.Bonus
FROM Sales.SalesPerson AS SP
INNER JOIN Person.Person AS P
    ON SP.BusinessEntityID = P.BusinessEntityID
INNER JOIN HumanResources.Employee AS E
    ON SP.BusinessEntityID = E.BusinessEntityID
WHERE E.MaritalStatus = 'S' 
ORDER BY Bonus DESC
```

# Objective:

find the person the Vice President of Engineering has starting dating and return their :

firstname

lastname

jobtitle

bonus

# Schema Needed:

Sales.SalesPerson

Person.Person

HumanResources.Employee

# NOTES:

if they have just starting dating their marital status should still be single

the bonus should be one of the top in the list

# Case 5

It is your job in WideWorldImporters to make sure that all orders are finalized and ready to ship out. Double check the orders and the order line to make sure there aren't any not ready.

```sql
USE WideWorldImporters

SELECT PurchaseOrderID
FROM  Purchasing.PurchaseOrders

INTERSECT

SELECT PurchaseOrderID
FROM Purchasing.PurchaseOrderLines

EXCEPT

SELECT PurchaseOrderID
FROM Purchasing.PurchaseOrders
WHERE IsOrderFinalized = 1
```

# Objective:

check the order IDs and return those whos orders are not finalized

# Schema Needed:

Purchasing.PurchaseOrders

Purchasing.PurchaseOrderLines

# NOTES:

order is either finalized or not 

it is given a 0 if it is not and a 1 if it is

# Case 6

There has been a nuclear fallout in the state of florida. You are part of a rescue team to save people registered in AdventureWorks2019. Find their addresses and go rescue them.

```sql
USE AdventureWorks2019

SELECT AddressLine1
FROM Person.StateProvince AS SP
INNER JOIN Person.Address AS A
    ON SP.StateProvinceID = A.StateProvinceID
WHERE SP.Name = 'Florida'
```

# Objective:

get the addresses of those affected in adventureworks2019

# Schema Needed:

Person.StateProvince

Person.Address

# NOTES:

Only those that live in florida are affected and are need of a rescue.

# Case 7

You are tasked to look over the new hires in AdventureWorks2019. Find out who they are and go show them around.

```sql
USE AdventureWorks2019

SELECT FirstName, LastName 
FROM Person.Person
WHERE BusinessEntityID IN 

(SELECT BusinessEntityID
FROM HumanResources.JobCandidate

INTERSECT

SELECT BusinessEntityID
FROM HumanResources.Employee
)
```

# Objective:

Find the names of those who were just hired and help them

# Schema Needed:

Person.Person

HumanResources.JobCandidate

HumanResources.Employee

# NOTES:

Only those with business entity IDs are counted as employees

Find those with IDs in the job candidate list
