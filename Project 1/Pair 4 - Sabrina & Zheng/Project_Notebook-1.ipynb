```sql
-- CASE #001: The Vanished Road Frame
-- Brief Case:
-- On July 31, 2013, a high-end bicycle component, the 'LL Road Frame – Black, 58',
-- disappeared from the AdventureWorks warehouse. A transaction exists, but inventory
-- is now zero. This case investigates the product trail and employee activity.
--
-- Objectives:
-- 1. Retrieve transaction details for ProductID 722 on '2013-07-31'.
-- 2. Review current inventory to confirm absence.
-- 3. Identify responsible employee via joins.
-- 4. Check communication logs for intent confirmation.
-- step 1:
USE AdventureWorks2017
GO
SELECT th.TransactionID, th.ProductID, p.Name AS ProductName, th.TransactionDate, th.Quantity
FROM Production.TransactionHistory th
JOIN Production.Product p ON th.ProductID = p.ProductID
WHERE th.ProductID = 722 AND th.TransactionDate = '2013-07-31';
```

```sql
SELECT *
FROM Production.ProductInventory
WHERE ProductID = 722;
```

```sql

SELECT e.BusinessEntityID AS EmployeeID, p.FirstName, p.LastName, e.JobTitle
FROM HumanResources.Employee e
JOIN Person.Person p ON e.BusinessEntityID = p.BusinessEntityID
WHERE e.BusinessEntityID = 277;
```

```sql
CREATE TABLE #CommunicationLogs (
    EmployeeID INT,
    Message VARCHAR(255),
    Date DATE
);

INSERT INTO #CommunicationLogs (EmployeeID, Message, Date)
VALUES 
(277, 'Frame 722 was logged. Let’s quietly take care of it.', '2013-07-31'),
(277, 'Don’t enter it into outbound records.', '2013-07-31');

SELECT *
FROM #CommunicationLogs
WHERE EmployeeID = 277 AND Date = '2013-07-31';
```
