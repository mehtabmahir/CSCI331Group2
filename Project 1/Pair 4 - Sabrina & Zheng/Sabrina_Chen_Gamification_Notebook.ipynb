# Gamification

<u>Case #1</u>

**Summary**

<span style="font-family: -apple-system, BlinkMacSystemFont, sans-serif; color: var(--vscode-foreground);">On 6/4/2011, a shipping truck with some of AdventureWorks' orders got into an accident. Some goods went missing or were irreparibly damaged in the accident. As such, orders from the involved products will need to be redone. Please identify the goods involved and relevant order details.</span>

<span style="font-family: -apple-system, BlinkMacSystemFont, sans-serif; color: var(--vscode-foreground);">(use Production.Product, Production.WorkOrderRouting)</span>

**Objectives**

- Retrieve all orders that were scheduled to begin shipment on 6/4/11
- Find orders that have discrepancies between their scheduled and actual start shipping dates
- Return the details of the goods/orders involved in the accident

**Schemas Needed to Solve the Case with Table or View Tables**

- <span style="font-family: -apple-system, BlinkMacSystemFont, sans-serif; color: var(--vscode-foreground);">Production.WorkOrder</span>
- <span style="font-family: -apple-system, BlinkMacSystemFont, sans-serif; color: var(--vscode-foreground);">Production.Product</span>

**Investigation Notes**

- Find the scheduled order dates of 6/4/11 that don't match up with the actual start date of the delivery of an order.
- Reutrn relevant details of the involved orders.

```sql
USE AdventureWorks2019

SELECT WOR.WorkOrderID, WOR.ProductID, WOR.ScheduledStartDate, WOR.ActualStartDate, WO.OrderQty
FROM Production.WorkOrderRouting as WOR, Production.WorkOrder as WO
WHERE WOR.ScheduledStartDate != WOR.ActualStartDate AND
      WOR.ScheduledStartDate = '2011-06-04'
```

<u>Case #2</u>

**Summary**

Some customers have complained that they haven't received the entirety of their orders. Identify which orders haven't been received in the total quantity requested so as to prioritize those orders to be finished.

**Objectives**

1. Find which orders' whose order quantity don't match with their received quantity
2. Return relevant details of each order found

**Schemas Needed to Solve the Case with Table or View Tables**

- Purchasing.PurchaseOrderDetail

**Investigation Notes**

- Compare orders whose OrderQty doesn't match their ReceivedQty
- Return details about each order

```sql
USE AdventureWorks2019

SELECT PurchaseOrderID, PurchaseOrderDetailID, ProductID, OrderQty, ReceivedQty, DueDate
FROM Purchasing.PurchaseOrderDetail
WHERE OrderQty != ReceivedQty
```
