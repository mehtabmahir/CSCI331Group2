## Case #001: The Payroll Leak

**Brief Case**

In March 2016, the finance team flagged multiple invoices with unusually low or negative profit margins. As part of a fraud detection initiative, your task is to

**Objectives -** 

1. Retrieve all invoice line items for March 2016.
    
2. Identify which products and customers were involved in low-profit sales.
    

1. Sort the results to find the **worst-performing or most discounted items**.
    

**Tables Needed:** AccessLogs, Employees, EmailCommunications

```sql
-- Sample: Find profit margins on invoice lines
SELECT InvoiceID, LineProfit
FROM Sales.InvoiceLines
ORDER BY LineProfit ASC;
```

## **ANSWER**

```sql
USE WideWorldImporters;
GO

SELECT 
    il.InvoiceID,
    il.StockItemID,
    il.Description,
    il.Quantity,
    il.UnitPrice,
    il.TaxRate,
    il.LineProfit,
    i.InvoiceDate,
    c.CustomerName
FROM Sales.InvoiceLines il
JOIN Sales.Invoices i ON il.InvoiceID = i.InvoiceID
JOIN Sales.Customers c ON i.CustomerID = c.CustomerID
WHERE 
    YEAR(i.InvoiceDate) = 2016 
    AND MONTH(i.InvoiceDate) = 3
ORDER BY il.LineProfit ASC;
```

## **Investigation Notes -**

- Queried invoice lines from `Sales.InvoiceLines` joined with `Sales.Invoices` to filter records from **March 2016**.
    
- Sorted the results by `LineProfit` in ascending order to identify **worst-performing sales**.
    
- Joined the data with `Sales.Customers` to associate each invoice with a specific customer.
    
- Extracted product descriptions, quantities, and unit prices to help audit the transactions.
    

## **Schemas -**

- Sales.InvoiceLines
    
- Sales.Invoices
    
- Sales.Customers

## Case #002: The Discount Scheme

**Brief Case**

AdventureWorks noticed that in January 2014, multiple high-value sales orders had unauthorized discounts applied, resulting in financial losses. Your job is to identify who applied the discounts.

**Objectives**

1. Retrieve sales orders with unusual discounts in January 2014.
2. Identify employees who authorized these discounts.
3. Review communication logs to find any collaboration between employees.

**Tables Needed:** SalesOrders, Discounts, Employees, CommunicationLogs

```sql
-- Sample SQL Query to detect unusual discounts
SELECT *
FROM SalesOrders
WHERE DiscountAmount > 500 AND OrderDate BETWEEN '2024-01-01' AND '2024-01-31';
```

## **ANSWER -**

```sql
USE AdventureWorks2019;
GO

-- Join orders with salespeople to find who authorized the discount orders
SELECT 
    soh.SalesOrderID,
    soh.OrderDate,
    sod.ProductID,
    sod.UnitPrice,
    sod.UnitPriceDiscount,
    p.FirstName + ' ' + p.LastName AS SalesPersonName,
    e.JobTitle
FROM Sales.SalesOrderHeader AS soh
JOIN Sales.SalesOrderDetail AS sod
    ON soh.SalesOrderID = sod.SalesOrderID
JOIN Sales.SalesPerson AS sp
    ON soh.SalesPersonID = sp.BusinessEntityID
JOIN HumanResources.Employee AS e
    ON sp.BusinessEntityID = e.BusinessEntityID
JOIN Person.Person AS p
    ON e.BusinessEntityID = p.BusinessEntityID
WHERE 
    YEAR(soh.OrderDate) = 2014 AND 
    MONTH(soh.OrderDate) = 1 AND
    sod.UnitPriceDiscount > 0;

```

## **Investigation Notes -**

- <span style="color: var(--vscode-foreground);">Queried </span> `Sales.SalesOrderHeader` <span style="color: var(--vscode-foreground);">and </span> `Sales.SalesOrderDetail` <span style="color: var(--vscode-foreground);">for all orders in </span> **January 2014** <span style="color: var(--vscode-foreground);">with a non-zero discount.</span>
- <span style="color: var(--vscode-foreground);">Calculated total price before and after discount to assess financial impact.</span>
- <span style="color: var(--vscode-foreground);">Joined discounted sales orders with the </span> `SalesPerson` <span style="color: var(--vscode-foreground);">table using </span> `SalesPersonID`<span style="color: var(--vscode-foreground);">.</span>
- Revealed that discounts were authorized by several **Sales Representatives**
- No table resembling `CommunicationLogs` exists in `AdventureWorks2019`.

## Case #003: The Fake Supplier Invoices

**Brief Case**  
AdventureWorks is reviewing product inventory transactions for the year 2014 to analyze how transactions impacted the available stock. You are tasked with identifying the total quantity of each product involved in transactions and comparing it with the current inventory levels.

**Objectives**

1. Retrieve all product transactions that occurred in 2014.
    
2. Calculate the total quantity change for each product from the transactions.
    
3. Check the inventory levels and compare them to the total quantity change.
    
4. Identify any discrepancies between the total transaction quantity and current inventory.
    

**Tables Needed**: `TransactionHistory`, `ProductInventory`

```sql
USE AdventureWorks2019;
GO

-- Sample query to get product transactions and current inventory quantities
SELECT 
    th.ProductID,
    th.TransactionDate,
    SUM(th.Quantity) AS TotalQuantityChanged,
    pi.Quantity AS CurrentQuantity
FROM Production.TransactionHistory AS th
JOIN Production.ProductInventory AS pi
    ON th.ProductID = pi.ProductID
WHERE th.TransactionDate BETWEEN '2014-01-01' AND '2014-12-31'
GROUP BY th.ProductID, th.TransactionDate, pi.Quantity
ORDER BY th.TransactionDate DESC;
```

## **Answer -**

```sql
USE AdventureWorks2019;
GO

-- Scalar function to calculate the total quantity change for a product
CREATE FUNCTION dbo.fn_GetTotalQuantityChange (@ProductID INT)
RETURNS INT
AS
BEGIN
    DECLARE @TotalQuantity INT;

    -- Calculate total quantity change for the given ProductID
    SELECT @TotalQuantity = SUM(Quantity)
    FROM Production.TransactionHistory
    WHERE ProductID = @ProductID;

    RETURN @TotalQuantity;
END;
GO

USE AdventureWorks2019;
GO

-- Query to retrieve transactions, related product information, and total quantity change for each product
SELECT 
    th.ProductID,
    th.TransactionID,
    th.TransactionDate,
    th.ReferenceOrderID,
    th.ReferenceOrderLineID,
    pi.Quantity AS CurrentQuantity,
    dbo.fn_GetTotalQuantityChange(th.ProductID) AS TotalQuantityChanged -- Using the scalar function
FROM Production.TransactionHistory AS th
JOIN Production.ProductInventory AS pi
    ON th.ProductID = pi.ProductID
WHERE 
    th.TransactionDate BETWEEN '2014-01-01' AND '2014-12-31'
ORDER BY th.TransactionDate DESC;



```

### **Investigation Notes**:

1. **Transaction Tracking**:
    
    - The `TransactionHistory` table contains records of product transactions with `ProductID`, `TransactionID`, `ReferenceOrderID`, and `ReferenceOrderLineID`.
        
    - Key quantities include `CurrentQuantity`, indicating the current stock after the transaction, and `TotalQuantityChanged`, which shows the total amount of quantity changed for each product.
        
2. **Discrepancy Analysis**:
    
    - Use the `ProductID` to track the total quantity change for products across multiple transactions, matching those changes with current inventory levels to check for discrepancies.
3. **Product Inventory**:
    
    - The data includes multiple `ReferenceOrderID`s and `ReferenceOrderLineID`s that are useful for linking transactions to specific orders and inventory changes.

### **Schema**:

- **ProductID**
    
- **TransactionID**
    
- **TransactionDate**
    
- **ReferenceOrderID**
    
- **ReferenceOrderLineID**
    
- **CurrentQuantity**
    
- **TotalQuantityChanged**

## Case #006: The Bonus Manipulation

**Brief Case**

AdventureWorks management suspects that bonuses were incorrectly processed in December 2011, resulting in certain employees receiving extra payments. You are to investigate.

**Objectives**

1. Retrieve bonus payment records for December 2011.
2. Identify employees who received irregular bonuses.
3. Check if the approving manager communicated with them about it.

**Tables Needed:** Bonuses, Employees, CommunicationLogs

```sql
-- Sample SQL Query to investigate bonuses
SELECT *
FROM Bonuses
WHERE PaymentDate BETWEEN '2011-12-01' AND '2011-12-31';
```

## **Answer -**

```sql
USE AdventureWorks2019;
GO

-- Step 1: Find pay rate changes in December 2011 (bonus investigation)
WITH BonusCandidates AS (
    SELECT eph.BusinessEntityID, 
           eph.Rate, 
           eph.RateChangeDate, 
           p.FirstName + ' ' + p.LastName AS EmployeeName,
           e.JobTitle
    FROM HumanResources.EmployeePayHistory AS eph
    JOIN HumanResources.Employee AS e
        ON eph.BusinessEntityID = e.BusinessEntityID
    JOIN Person.Person AS p
        ON e.BusinessEntityID = p.BusinessEntityID
    WHERE 
        YEAR(eph.RateChangeDate) = 2011 AND 
        MONTH(eph.RateChangeDate) = 12
)

-- Step 2: Simulate manager list and cross join with bonus candidates
SELECT 
    bc.BusinessEntityID,
    bc.EmployeeName,
    bc.JobTitle,
    bc.Rate,
    bc.RateChangeDate,
    m.BusinessEntityID AS ManagerID,
    m.FirstName + ' ' + m.LastName AS ManagerName,
    m.JobTitle AS ManagerTitle
FROM BonusCandidates bc
CROSS JOIN (
    SELECT 
        e.BusinessEntityID, 
        e.JobTitle, 
        p.FirstName, 
        p.LastName
    FROM HumanResources.Employee e
    JOIN Person.Person p 
        ON e.BusinessEntityID = p.BusinessEntityID
    WHERE e.JobTitle LIKE '%Manager%' OR e.JobTitle LIKE '%Director%'
) AS m;
```

## **Investigation Notes** 

  

- Queried `EmployeePayHistory` for rate changes in **December 2011**.
    
- Identified one significant anomaly:
    
    - **Rob Walters**, **Senior Tool Designer**, received a rate of **29.8462**, significantly above others at 7.25.
        
- Queried the `Employee` and `Person` tables to identify employees with managerial job titles.
    
- Listed potential managers who may have authorized or overlooked the pay change:
    
    - Notable managers included **Roberto Tamburello (Engineering Manager)**, **Dylan Miller (R&D Manager)**, and **Peter Krebs (Production Control Manager)** among others.

## **Case #007: The Unauthorized Price Changes**

**Brief Case**

<span style="color: var(--vscode-foreground);">In </span> **October 2017**<span style="color: var(--vscode-foreground);">, several structural changes were made to core dimension tables in the </span> `AdventureWorksDW2019` <span style="color: var(--vscode-foreground);"> data warehouse, </span> <span style="color: var(--vscode-foreground);">These changes could affect reporting, ETL processes, and other dependent systems. You are tasked with investigating who made these changes, which tables were affected, and whether any user modified multiple tables during that time.</span>

**Objectives**  

- Identify all structural  changes made in October 2017.
    
- Determine which systems were affected by the changes.
    
- Identify the user(s) who performed the changes and investigate whether the same user altered multiple tables.
    

**Tables Needed:** DatabaseLog

```sql
-- Check for schema changes (ALTER_TABLE) in the DatabaseLog for October 2017
SELECT Event, [Schema], Object, PostTime
FROM DatabaseLog
WHERE Event = 'ALTER_TABLE' 
  AND YEAR(PostTime) = 2017 
  AND MONTH(PostTime) = 10;
```

## **Answer -**

```sql
USE AdventureWorksDW2019;
GO

-- Drop the existing function if it exists
IF OBJECT_ID('dbo.fn_GetSchemaChangesByMonth', 'FN') IS NOT NULL
    DROP FUNCTION dbo.fn_GetSchemaChangesByMonth;
GO


USE AdventureWorksDW2019;
GO

-- Create the Inline Table-Valued Function (TVF)
CREATE FUNCTION dbo.fn_GetSchemaChangesByMonth (@Year INT, @Month INT)
RETURNS TABLE
AS
RETURN
    SELECT *
    FROM dbo.DatabaseLog
    WHERE 
        Event = 'ALTER_TABLE'
        AND YEAR(PostTime) = @Year
        AND MONTH(PostTime) = @Month;
GO


-- Use the TVF to retrieve schema changes for October 2017
SELECT * 
FROM dbo.fn_GetSchemaChangesByMonth(2017, 10);

USE AdventureWorksDW2019;
GO

-- Step 1: Create the Common Table Expression (CTE)
WITH SchemaChanges AS (
    SELECT 
        DatabaseLogID,
        DatabaseUser,
        [Schema],
        Object,
        Event,
        PostTime,
        TSQL
    FROM dbo.DatabaseLog
    WHERE 
        Event = 'ALTER_TABLE'
        AND YEAR(PostTime) = 2017
        AND MONTH(PostTime) = 10
)

-- Step 2: Identify the user with the most table changes
SELECT 
    DatabaseUser,
    COUNT(DISTINCT Object) AS TablesChanged,
    STRING_AGG(Object, ', ') AS ModifiedTables,
    MIN(PostTime) AS FirstChange,
    MAX(PostTime) AS LastChange
FROM SchemaChanges
GROUP BY DatabaseUser
ORDER BY TablesChanged DESC;


```

### **Investigation Notes**

- The primary focus is on the **schema changes** (`ALTER_TABLE`) in dimension tables, specifically looking for **potential unapproved changes** that could impact reports or processes.
    
- The changes occurred in **October 2017**, and the goal is to pinpoint which **users** were responsible for modifying table structures in the **`AdventureWorksDW2019`** database.
    
- All activities were logged in the **`DatabaseLog`** table, with schema changes being captured as **`ALTER_TABLE` events**.
    
- No data-level changes were made (e.g., no updates or deletes), just structural changes. The analysis will involve identifying the **users** responsible for the changes and whether these changes were **isolated** or part of a **larger pattern**.
